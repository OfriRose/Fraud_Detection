# Exploratory review of generated reports

> **Notebook status:** The modules in `src/fraud_detection/` and the training/scoring CLIs are authoritative. This notebook is a thin explanatory view; old embedded outputs, ad hoc logic, and stale claims were removed.

`split_summary.csv` documents the chronological train/validation/test protocol. The monthly and category tables are **test-set diagnostic slices** generated after evaluation. They are useful for monitoring heterogeneity, but they must not be used to retune the model or its already-locked threshold.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

report_dir = Path("reports")
split_summary = pd.read_csv(
    report_dir / "split_summary.csv",
    parse_dates=["date_start", "date_end"],
)
monthly_test = pd.read_csv(report_dir / "test_monthly_metrics.csv")
category_test = pd.read_csv(report_dir / "test_category_metrics.csv")
```

```python
display(split_summary)
display(monthly_test)
display(category_test.sort_values("transactions", ascending=False).head(10))
```

Read prevalence, precision, recall, review rate, and cost together: small or all-positive slices can produce extreme-looking metrics. These tables are descriptive evidence from the held-out test period, not independent model-selection experiments.
